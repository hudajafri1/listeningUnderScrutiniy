####Background Information:

Users often reference alternative platforms and forms of music consumption. Unlike Spotify, which operates primarily on a streaming access model, these platforms are often associated with more direct forms of artist support. Additionally, physical formats often come up in conversation as well, such as iPods, CDs, and vinyl records. These forms of music consumption are tied to ownership rather than access, allowing users to retain permanent copies of music rather than relying on a subscription.

##Network Analysis

To understand user interactions on Bluesky, I constructued a directed reply network where each node represents a Bluesky user and each edge represents a reply from one user to another. This allowed me to analyze conversational structure and note if discussions form through isolated clusters or interconnected communities. I used the given Python script to convert this into a GraphML file.

![gephigraph](https://drive.google.com/uc?id=1dKSE31fUa-YklM9Z1NVNYB4trTRYLWrp)

My main tool was Force Atlas 2, which I used repeatedly between steps to  spatialize the data. I scaled the node size according to the replies, with a range of 5 to 500. Using the modularity class settings, I colored the nodes. I focused on the largest nine communities, which together accounted for 8.56% of edges and 3.84% of the nodes. This allowed me to highlight the most active connversational clusters.

I found the major contributors to be 50501: The People's Movement (50 protests, 50 states, 1 day), David (normal user), and Sustained Decay (SF CA Bay Area weirdo passionate about the arts but especially synth sound). There seem to multiple conversations that aren't necessarily interconnected, as they don't extend across communities.

There does seem to be a lot of conversation regarding Spotify, but given how little the groups interact with each other, there seems to be an echo chamber going on with content circulating within communities rather than across them. This implies Bluesky exists more as a way to vent frustrations with Spotify rather than organize a movement, so interest in owning music seems to be more of a niche response rather than a dominant shift.

###Exploration of Highly Active Users:
I wanted to explore the participation of highly active users based on how many replies they received versus how many they made. Through this, I was able to differentiate between “attention hubs” and “active participants.” Some accounts, such as 50501movement.bsky.social and indivisible.org, received a high number of replies but did not actively engage with others, suggesting they act as focal points in the conversation. In contrast, users like xmissedthespot.bsky.social and caillan74.bsky.social contributed many replies but received relatively little engagement, indicating they play a more participatory role. Additionally, the presence of domains such as forbes.com and indivisible.org suggests that replies are not only directed at individuals but also at shared content, highlighting the role of external media in shaping the Spotify discourse on Bluesky.

```python
import pandas as pd

df = pd.read_csv("newData.csv")
did_to_handle = dict(zip(df['author_did'], df['author_handle']))
import networkx as nx

G = nx.DiGraph()
for _, row in df.iterrows():
    author = row['author_handle']
    parent_did = row['reply_parent_author_did']

    if pd.notna(parent_did) and parent_did in did_to_handle:
        parent = did_to_handle[parent_did]
        G.add_edge(author, parent)
in_degree = dict(G.in_degree())
top_replied = sorted(in_degree.items(), key=lambda x: x[1], reverse=True)[:10]
out_degree = dict(G.out_degree())


top_active = sorted(out_degree.items(), key=lambda x: x[1], reverse=True)[:10]
df_in = pd.DataFrame(top_replied, columns=['user', 'replies_received'])
df_out = pd.DataFrame(top_active, columns=['user', 'replies_made'])
comparison = pd.merge(df_in, df_out, on='user', how='outer')
print(comparison)
```

#Sentiment Analysis Using TextBlob
Early posts tend to lean more positive or neutral in tone, and are often a result of music recommendations. As time goes on, the average sentiment dips below the zero threshold, specifically in late 2025. This indicates growing frustration and criticism among users. This downward trend may be driven by posts addressing Spotify’s platform issues, such as algorithm complaints, intrusive ads, or dissatisfaction with playlist recommendations. Spikes in negativity often coincide with highly replied-to posts, suggesting that controversial or critical messages generate strong engagement and further amplify negative sentiment. While some highly active users continue to contribute positive or neutral replies, the overall conversation reflects a shared frustrated reaction.

```python
from textblob import TextBlob
import pandas as pd
import matplotlib.pyplot as plt

def get_sentiment(text):
    try:
        return TextBlob(str(text)).sentiment.polarity
    except:
        return 0

df['sentiment'] = df['text'].apply(get_sentiment)

# Fix datetime
df['created_at'] = pd.to_datetime(df['created_at'], format='mixed', errors='coerce')
df = df.dropna(subset=['created_at'])

# Filter to 2024 onward
df_recent = df[df['created_at'] >= '2024-01-01']

# Group
sentiment_over_time = df_recent.groupby(df_recent['created_at'].dt.date)['sentiment'].mean()

# Plot
sentiment_over_time.plot()
plt.title("Sentiment of Spotify Discussion Over Time (2024–Present)")
plt.xlabel("Date")
plt.ylabel("Average Sentiment")
plt.axhline(0, color='red', linestyle='--', label='Neutral')
plt.legend()
plt.tight_layout()
plt.show()
```

###Findings

####Memo 1: Ethical Framing of Spotify Through Artist Compensation
Going into this project, I knew the price increase would be a point of contention for Spotify users. I initially believed this was primarily for “selfish” reasons, as in frustration over an increase in spending. However, thematic analysis led me to conclude that criticism is framed less around personal inconvenience and more around ethical concerns about artist compensation. In fact, the price increase often intensifies the concern, as users question why they are being charged more despite artists still being paid so little per stream.

One user writes, “When I learned that artists were not compensated fairly on Spotify, that only made me more stubborn about purchasing CDs and ripping them.” Another compares payouts across platforms, saying, “Qobuz pays $0.021$ per stream … Spotify pays a measly $0.005.” When considered alongside the rising subscription costs, this comparison reinforces the perception that Spotify is extractive rather than supportive of artists. Even users who continue to use Spotify do not necessarily view it in a positive light, as reflected in the post, “Spotify is in the mix, but prefer to contribute to making artists actual money.” Such responses indicate that an increased price does not translate to perceived fairness, but rather, sharpens ethical scrutiny.

The general conversation seems to be shifting away from streaming as a convenience and toward the ethics of participation. Bluesky discussion around Spotify, even among active users, reflects an ongoing moral negotiation about whether using the platform aligns with their values. This represents a shift in how users conceptualize digital platforms. Spotify is no longer seen as a neutral service for streaming, but rather as an entity responsible for how value is distributed within the music industry. The emphasis on per-stream payouts shows users engaging with the infrastructure of the music industry beyond just the music they consume.

In addition, this theme reveals a tension between ethical awareness and behavioral change. Based on posts, though some users cancel subscriptions or switch platforms, many continue to use Spotify. In these cases, Bluesky is used as a place to express frustration as opposed to a place to organize a boycott. This implies that convenience and entrenched usage patterns can outweigh ethical concerns. The discourse becomes more expressive than transformative, contributing to a shared education around Spotify’s practices without a fundamental alteration of usage patterns. Ultimately, this theme shows how digital consumption is becoming increasingly moralized, with Spotify serving as a focal point for broader concerns about fairness in creative industries.


####Memo 2: Identity Surrounding Non-Use of Spotify
Another recurring theme is the way users frame their lack of Spotify use, whether this be through never having the app or having abandoned it, as a form of identity performance. Rather than offering alternatives or engaging deeply with critiques, many posts emphasize personal distance from the platform as a kind of “flex.”

Examples of this are especially explicit. One user states, “I would never ever use Spotify, never have, never will.” Another similarly says, “I’ve never had a Spotify subscription… I’m good.” A more aggressive version frames usage as a kind of failure: “It’s amazing how people will contort themselves into knots to insist they need to use Spotify… if I see you posting Spotify Wrapped I’m blocking you.” Rather than inviting discussion as expected on a social media site, these posts position the speaker as already outside the issue, therefore, above it.

Even when alternatives are mentioned, they are often presented briefly or almost as an afterthought. For example, “Sign up for Tidal, Qobuz, Bandcamp, CDs, vinyl…” While this lists options, it does not engage with the practical challenges of switching or explain why these alternatives might actually work better in everyday use. The suggestion feels more like a gesture than a fully developed recommendation.

This theme is interesting because it shows how discourse can shift from systemic critique to self-positioning. The focus moves away from Spotify’s practices and toward the individual’s relationship to the platform. Not using Spotify becomes less about solving a problem and more about signaling taste, ethics, or independence from mainstream platforms.
This is significant because it can limit the conversation from turning into a movement. While these posts may raise awareness in a minimal sense, they often shut down dialogue rather than expanding it, and lack practical solutions. By framing non-use as an obvious or easy choice, they overlook the structural reasons why Spotify remains dominant, including convenience, social integration, and discovery features, and further alienate those stuck in the habit of usage.

At the same time, this behavior reflects broader patterns in online culture, where consumption choices are used to signal identity. Avoiding Spotify becomes similar to rejecting certain brands or platforms as a way of expressing values. This creates an exclusionary tone, especially when users imply that those who remain on Spotify are uninformed or unwilling to make ethical choices. Ultimately, there is a tension between critique and accessibility. While rejecting Spotify may feel like a strong ethical stance, presenting it as a simple personal choice can flatten the complexity of the issue and discourage more nuanced or constructive discussion.


####Memo 3: Recommendation Culture as Community
Another emerging theme is the emphasis on alternative ways of consuming music, often framed as both more ethical and more intentional. Unlike the “flex” posts shown in Memo 2, these contributions tend to describe specific practices, such as using different platforms, buying physical media, or combining multiple sources for listening.

For example, one user explains, “I recently ended my Spotify subscription. Now I’m using Qobuz… Bonus for the great artists.” Another describes a mixed approach: “Apple Music for driving… YouTube / Bandcamp behind that… Sometimes I’ll scrape Spotify to find new stuff.” A third highlights physical formats: “For us, it’s vinyl, tapes, CDs and Bandcamp.” These posts provide more concrete insight into how users actually navigate music consumption outside of, or alongside, Spotify. Rather than presenting a single replacement, these posts illustrate a hybrid model where different platforms work for different purposes, such as discovery, convenience, or artist support.

This theme is significant because it demonstrates that alternatives do exist, but they are often fragmented and require active effort to maintain. No single platform fully replaces Spotify’s convenience, which leads users to piece together their own personalized systems. For instance, a user may use Spotify for discovery, Bandcamp for purchasing music, and physical media for long-term ownership. This can be empowering, since it allows for more intentional support of artists and greater control over listening habits. At the same time, it requires more effort and knowledge, which may limit the scale at which these practices are adopted. Therefore, though these recommendations are more practical than those in Memo 2, they are still not necessarily accessible.

At the same time, these practices highlight the limits of individual action. While users can choose to support artists more directly, the overall structure of the streaming economy remains largely unchanged. This raises questions about scale and whether these alternative practices can meaningfully challenge Spotify’s dominance or if they remain relatively niche. Overall, this theme offers a more nuanced and constructive side of the discourse. It moves beyond critique or identity performance and instead focuses on what users are actually doing. In doing so, it reveals both the possibilities and the constraints of navigating music consumption within a platform-dominated landscape.
